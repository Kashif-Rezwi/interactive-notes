#**PART 1: Mathematical Foundations & Notation**


###**Scalar**


A scalar is a single numerical value.
- a ∈ R

**Examples:**
- Learning rate
- Weight value
- Bias term
- Loss value

In ML systems, scalars are often used as optimization parameters or evaluation metrics.

###**Vector**



A vector is an ordered collection of numbers arranged in a single dimension.

- $$x = [x_1,x_2,...,x_n]^T ∈ R^n$$


**A vector represents:**
  - Feature vectors
  - Word embeddings
  - User representations
  - Probability distributions
  - Hidden states in neural networks

Example:
$$x=
\begin{bmatrix}
170 \\
65\\
25
\end{bmatrix}
$$


**This could represent:**
- Height
- Weight
- Age

of a person in a dataset.

###**Matrix**


A matrix is a 2-dimensional arrangement of numbers.
 $$A ∈ R^{m * n}$$

**Where:**


- m = number of rows
- n = number of columns

**In ML:**
- Matrix Dimension	Interpretation
- Rows	Samples / Observations
- Columns	Features

**Example:**

$$A=
\begin{bmatrix}
1 & 2 \\
3 & 4\\
5 & 6
\end{bmatrix}
$$


**This represents:**
- 3 samples
- 2 features

Datasets in ML pipelines are almost always represented as matrices.

$$f:R^n

**This means:**

- Input → n-dimensional vector
- Output → single real value

**Example: Linear Model**
$$f(x)=w

**Where:**

- w = weights
- x = feature vector
- b = bias

**This equation forms the foundation of:**
- Linear Regression
- Logistic Regression

Every neural network layer is essentially a sequence of linear transformations followed by non-linear activations.

###**Summation Notation**

Summation notation provides a compact mathematical representation.


$$
\sum_{i=1}^{n} x_i
$$


**Meaning:**
- Add all elements from index 1 to n

**Used extensively in:**
- Loss functions
- Gradient calculations
- Optimization equations
- Statistical computations


**Example:**

$$\sum_{i=1}^{4}x_i = x_1 + x_2 + x_3 + x_4$$


###**Important Notation**


- $$x^T → transpose$$
- $$||x|| → norm$$
- $$Rn → n-dimensional space$$


###**Proof Intuition (Interview Skill)**



Interviewers often test conceptual understanding through small mathematical proofs.


**Example: Linearity of Dot Product**
$$(x+y)⋅z=x⋅z+y⋅z$$

Expanding using summation notation:

$$(x+y)⋅z=\sum_{i=1}^{n}(x_i + y_i)z_i $$
	​
	​


Distributing:
=$$\sum_{i=1}^{n}(x_i z_i) + \sum_{i=1}^{n}(y_i z_i)$$
	​


Therefore:

$$(x+y)⋅z=x⋅z+y⋅z$$


**This property is known as linearity.**

Such derivations are extremely important in:

- Gradient derivations
- Backpropagation proofs
- Optimization theory
- ML interviews

#**PART 2: VECTOR SPACES & GEOMETRY**


**Vector Space Definition**

- A vector space is a mathematical structure

where:
- Vector addition is valid
- Scalar multiplication is valid

If vectors remain inside the same space after these operations, the set forms a vector space.

**Example:**

$$R^2$$


contains all 2D vectors.

**Linear Combination**

A linear combination is formed by scaling vectors and adding them together.

$$a_1 x_1+a_2	x_2+...+a_kx_k$$
	​


**This concept is central to:**
- Feature engineering
- Neural network representations
- Latent embeddings
- PCA

Every prediction in a linear model is essentially a weighted linear combination of features.

###**Span**

- The span of vectors is the set of all possible linear combinations.
If vectors can generate an entire space, they are said to span that space.

**Example:**
- Two non-parallel vectors in 2D span the entire plane.

**In ML:**
- Feature representations span information space
- Embeddings span semantic spaces

###**Basis**

A basis is the minimum set of vectors required to span a space.

**Properties:**
- Vectors must be linearly independent
- No redundant vectors allowed

**Example:**
$${(1,0),(0,1)}$$

forms the standard basis of $$R^2$$


###**Dimension**

- Dimension is the number of basis vectors required to represent a space.

**Examples:**
- Line → dimension 1
- Plane → dimension 2
- Space → dimension 3


**Modern ML systems often operate in:**
- Hundreds of dimensions
- Thousands of dimensions
- Millions of dimensions
