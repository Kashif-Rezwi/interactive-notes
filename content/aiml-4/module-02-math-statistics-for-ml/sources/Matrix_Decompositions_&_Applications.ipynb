##**Eigenvalues & Eigenvectors**

**Suppose a matrix transformation acts on vectors.**

**Normally:**
- vectors rotate
- change direction
- stretch/shrink

**But some special vectors behave differently.**

**These vectors:**
- keep the SAME direction
- only change magnitude

**These are called:**
- Eigenvectors
- Eigenvalues

**Mathematical Definition**

For a square matrix:
 $$Av=λv$$

**Where:**
- A = matrix
- v = eigenvector
- λ = eigenvalue

**Example**

Suppose:
$$A=
\begin{bmatrix}
2 & 0 \\
0 & 3\\
\end{bmatrix}$$


**Transformation:**
- x-direction scaled by 2
- y-direction scaled by 3

**Eigenvectors:**
- $$[1,0]$$
- $$[0,1]$$

**Eigenvalues:**
- 2
- 3



####**Important ML Intuition**

**Eigenvectors reveal:**
- dominant directions
- stable patterns
- intrinsic geometry of data

**This is foundational for:**
- PCA
- Spectral methods
- Dynamical systems
- Covariance analysis

$${λ_1,λ_2,…,λ_n}$$

##**Diagonalization**

**Convert a matrix into diagonal form:**
$$A=PDP^{-1}$$

**Where:**
- P = eigenvector matrix
- D = diagonal eigenvalue matrix



**Why Is Diagonalization Useful?**
- Because diagonal matrices are EASY to compute.

**Example:**
$$A^{100}=PD^{100}P^{-1}$$

- Instead of multiplying 100 times.


**Computational Advantage**
**Operations become easier:**
- Matrix powers
- Exponentials
- Differential equations
- Dynamical systems

####**Matrix Powers**

**Instead of:**
$$A^{100}$$


**compute:**
$$A^{100}=PD^{100}P^{-1}$$

**Since:**
$$D^{100}$$

only raises diagonal elements individually.

**Diagonalizability Condition**

- A matrix is diagonalizable iff:

  - Geometric multiplicity=Algebraic multiplicity

**Equivalent statement:**
- matrix has n linearly independent eigenvectors

####**ML Insight**

**Diagonalization helps:**
- understand transformations
- simplify covariance structures
- analyze stability
- interpret neural network dynamics

##**Singular Value Decomposition (SVD)**

**Most Important Matrix Decomposition in ML**

**SVD works for:**
- ALL matrices
- rectangular matrices
- non-square matrices

Unlike eigen decomposition.

**SVD Formula**

$$A=UΣV^T$$


**Where:**
- U → left singular vectors
- Σ → singular values
- V^T→ right singular vectors

**Deep Geometric Interpretation**

SVD decomposes transformation into:


**Step 1**
- Rotation/reflection

**Step 2**
- Axis-wise scaling

**Step 3**
- Another rotation/reflection

Thus every linear transformation can be interpreted geometrically.

**Singular Values**

**Singular values indicate:**
- importance of directions
- energy captured
- information strength

**Larger singular value:**
- → more important structure.

**Smaller singular value:**
- → likely noise.



#####**Why SVD Is Extremely Powerful**
**1. Noise Reduction**
- Remove tiny singular values.

**2. Compression**
- Store only important components.

**3. Latent Feature Extraction**
- Find hidden structure in data.

**4. Numerical Stability**
- Widely used in scientific computing.



**Connection to Eigenvalues**

Singular values are:

$$σ_i = \sqrt{(λ_i(A^TA)}$$

##**PCA Connection (VERY IMPORTANT)**

**Goal of PCA**

**Find directions with:**
- maximum variance
- minimum redundancy

PCA projects high-dimensional data into fewer dimensions.







**Covariance Matrix**

- PCA starts from covariance matrix:


$$\frac{1}{n}X^TX$$



**PCA computes:**
- eigenvectors of covariance matrix
- largest eigenvalues

**Principal Components**
- Principal components represent:
  - directions of highest variance
  - most informative dimensions

**MOST IMPORTANT INSIGHT**
- PCA is fundamentally SVD on centered data.
**If:**
- data is mean-centered

Then:
$$X=UΣV^T$$


**The columns of:**
- V = principal directions
- singular values = variance information







#####**Core PCA Insight**

**Principal components are:**

- eigenvectors of covariance matrix
- ordered by descending eigenvalues

**PCA Workflow**

Step 1
- Center data

Step 2
- Compute covariance matrix

Step 3
- Compute eigenvectors/SVD


Step 4
- Keep top-k components

Step 5
- Project data



**Real ML Applications**

**Computer Vision**
- Face recognition

**NLP**
- Word embedding reduction

**Finance**

- Risk factor analysis

**Bioinformatics**

- Gene expression compression

**MOST IMPORTANT CONNECTION**

- If centered data matrix:

$$X=UΣV^T$$

**Then:**

- columns of V = principal directions
- singular values encode explained variance

**Thus:**

- PCA = SVD on centered data

##**Low-Rank Approximation**


- Approximate matrix using only top-k singular values.

**Instead of full matrix:**
$$A=UΣV^T$$

**Use:**

$$A_k=U_k	Σ_k V^T_k$$
	​


**Why Is This Powerful?**

Because most real-world data has:
- redundancy
- hidden structure
- correlated features

**Meaning:**
- large matrices can often be approximated using few components.  



**Approximation Error**

- Error depends on discarded singular values.

**Lower discarded singular values:**
- → better approximation.
